### Exercicio 1
Em um cercado existem vários patos e coelhos. Escreva uma função em linguagem 
Python que recebe como parâmetro o total de cabeças e o total de pés, e retorne dois 
valores: o total de patos e o total de coelhos que se encontram nesse cercado. (4,0 
pontos) 
Teste essa função elaborando um script que vai receber do usuário dois valores. Para 
tanto, leve em consideração que os dois últimos dígitos de seu RA sejam iguais a Y. 
 
Entrada de Dados:    
Total de Cabeças:  Y = _________ 
 
Total de Pés: Y*3 = _________ 
 
Levando em consideração o resultado de seu script... Responda: 
 
Total de Patos:___________________        Total de Coelhos: ___________________

```python
def total(cabeca,pe):
    c = cabeca
    p = pe
    c *= -2 #c=c*-2
    c += pe
    c //= 2
    p = cabeca - c
    if(pe%2==0):
        print(f"Total de Patos: {p}\nTotal de Coelhos: {c}")
    else:
        print(f"O resultado inteiro não é possivel\nResultado aproximado\nTotal de Patos: {p}\nTotal de Coelhos: {c}")
    
```

```python
total(11,33)
```

### Exercio 2
 
Faça um script em Python que recebe como entrada um número positivo no formato de 
String. Esse script deve devolver dois valores: o total da soma de todos os dígitos desse 
número e o valor da multiplicação de todos esses mesmos dígitos. Por exemplo: Se a 
entrada for 1234, o script deve devolver como resultado: Soma = 10 e Multiplicação = 
24.       (3,0 pontos) 
Passe o número completo de seu RA para o script, e responda: 
 
RA = _____________________________ 
 
Multiplicação = _____________________ 
 
Soma = ____________________________

```python
def soma_multi(ra):
    soma = 0
    multi = 1
    for i in ra:
        soma += int(i)
        multi *= int(i)
    print(f"""
    Ra = {ra}
    Soma = {soma}
    Multiplicação = {multi}
    """)
```

```python
ra = input("Digite seu RA")
soma_multi(ra)
```

### Exercicio 3

Reverso do número. Faça uma função que retorne o reverso de um número inteiro 
informado. Por exemplo: 127 -> 721.   ((use os números de seu RA para testar))

```python
def reverso(ra):
    ra_string = str(ra)
    ar = ""
    for i in ra_string:
        ar = i+ar
    return ar    
```

```python
ra = int(input("Digite um numero: "))
print(f"O reverso desse numero é {reverso(ra)}")
```
