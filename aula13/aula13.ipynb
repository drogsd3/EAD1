Crie uma função chamada eh_par(numero) que receba um número inteiro como parâmetro e retorne True se ele for par e False caso contrário.Depois, peça para o usuário digitar um número e mostre a mensagem apropriada.

```python
def eh_par(n):
    if n % 2 == 0:
        return True
```

```python
def eh_par1(n):
    if n % 2 == 0:
        return 'par'
    return 'impar'
```

```python
n = int(input("Digite um número: "))
if eh_par(n):
    print("O número é par!")
else:
    print("O número é ímpar!")
```

Implemente a função calcular_imc(peso, altura) que recebe o peso (kg) e a altura (m), e retorna o valor do IMC. Depois, crie uma segunda função classificar_imc(imc) que classifica o resultado como:
- Abaixo do peso (< 18.5)
- Peso normal (18.5 a 24.9)
- Sobrepeso (25 a 29.9)
- Obesidade (≥ 30)

```python
def calcular_imc(p,a):
    return p / a**2
```

```python
def classificar_imc(imc):
    if imc < 18.5:
        return 'Abaixo do peso'
    elif imc <= 24.9:
        return 'Peso normal'
    elif imc < 30:
        return 'Sobrepeso'
    else:
        return 'Obesidade'
```

```python
p = float(input("Peso (kg): "))
a = float(input("Altura (m): "))
imc = calcular_imc(p, a)
print(f"IMC: {imc:.2f} - {classificar_imc(imc)}")
```

```python
def eh_primo(n):
    if n <= 1:
        return False
    else:
        max = int(n ** (1/2))
        for i in range(2, max+1):
            if (n % i) == 0:
                return False
        return True
```

```python
numero = int(input("Digite um numero: "))
if eh_primo(numero):
    print("O numero é primo")
else:
    print("O numero não é primo")
```
