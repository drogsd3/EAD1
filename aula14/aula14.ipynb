### Funções com Parâmetro Padrão

```python
def calcular_entrega(peso, regiao, presente = False):
    """
    DocString usando as 3 aspas dupla ou simples, para documentação e entendimento do codigo

    """
    valor_entrega = 0
    
    if peso <= 500:
        valor_entrega += 15
    elif peso <= 2000:
        valor_entrega += 43
    else:
        valor_entrega += 103
        
    if regiao == 'N' or regiao == 'Ne':
        valor_entrega += 50
    if regiao == 'Co' or regiao == 'S':
        valor_entrega += 25

    if presente:
        valor_entrega += 10

    return valor_entrega
```

```python
def regiao(destino):
    norte = ["AC", "AP", "AM", "PA", "RO", "RR", "TO"]
    nordeste = ["AL", "BA", "CE", "MA", "PB", "PE", "PI", "RN", "SE"]
    centro_oeste = ["DF", "GO", "MT", "MS"]
    sudeste = ["ES", "MG", "RJ", "SP"]
    sul = ["PR", "RS", "SC"]
    
    if destino.upper() in norte:
        return 'N'
    if destino.upper() in nordeste:
        return 'Ne'
    if destino.upper() in centro_oeste:
        return 'Co'
    if destino.upper() in sul:
        return 'S'
    if destino.upper() in norte:
        return 'Se'
    return 'NA'
```

```python
peso = float(input('Digite o peso do produto em gramas.'))
destino = input('Digite o Estado da entrega. Ex: SP se for São Paulo.')
destino = regiao(destino)
presente = input('É presente? S - Sim ou N - Não.')
if presente.lower() == 's':
    presente = True
else:
    presente = False
valor_entrega = calcular_entrega(peso, destino, presente)

print(f'O valor do frete da entrega é R${valor_entrega:.2f}')
```

Faça um jogo completo “pedra – papel – tesoura”. <br>
A pessoa joga contra o computador. 
<br>Ao iniciar, pergunta-se o nome do jogador.
<br>Ao final de cada Jogo, pergunta se quer continuar jogando.
<br>Ao finalizar, depois de “n” jogadas, apresenta-se o resultado:
- Vitórias de <nome>....: 3
- Vitórias do Computador: 2
- Empates...............: 4
- <nome> foi o grande Campeão!

```python
nome = input("Digite seu nome: ")
escolha = ['Pedra', 'Papel','Tesoura']
vitoria_jogador = 0
vitoria_computador = 0
empates = 0
ej = 0
ec = 0
while True:
    escolha_jogador = input('Escolha Pedra, Papel ou Tesoura')
    escolha_computador = choice(escolha)
    
    if escolha_jogador.lower() == 'pedra':
        ej = 0
    elif escolha_jogador.lower() == 'papel':
        ej = 1
    elif escolha_jogador.lower() == 'tesoura':
        ej = 2
    else:
        print("Escolha inválida! Tente novamente.")
        continue # Volta para o início do while se o usuário digitar errado

    if escolha_computador.lower() == 'pedra':
        ec = 0
    elif escolha_computador.lower() == 'papel':
        ec = 1
    elif escolha_computador.lower() == 'tesoura':
        ec = 2

    resultado = (ej - ec) % 3
    print(f'Você escolheu {escolha_jogador}\nMaquina escolheu {escolha_computador}')
    if resultado == 0:
        print('Empate')
        empates += 1
    if resultado == 1:
        print('Vitoria')
        vitoria_jogador += 1
    if resultado == 2:
        print('Derrota')
        vitoria_computador += 1
        
    para = input('Deseja continuar? S - Sim ou N - Não')    
    if para.upper() == 'N':
        break

print(f"""
- Vitórias de {nome}....: {vitoria_jogador}
- Vitórias do Computador: {vitoria_computador}
- Empates...............: {empates}
""")
if vitoria_jogador > vitoria_computador:
    print(f"{nome} foi o grande Campeão!") 
elif vitoria_computador > vitoria_jogador:
    print("O Computador foi o grande Campeão!")
else:
    print("O jogo terminou em um empate geral!")
```
