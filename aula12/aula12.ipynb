```python
from random import randint
dado = []
for i in range(6000):
    dado.append(randint(1,6))

for i in range(1,7):
    print(f"Recorrencia do {i}: {dado.count(i)}")
```

```python
numeros = [0,0,0,0,0,0,0]
for i in range(6000):
    n = randint(1,6)
    numeros[n] = numeros[n] + 1
for i in range(1,7):
    p = (numeros[i] / 6000) * 100
    print(f"{i} - {p:6.2f}")
```
